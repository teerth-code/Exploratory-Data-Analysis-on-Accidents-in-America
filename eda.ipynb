```
df.isna().sum().sort_values(ascending = False)
```

```
percentage_missing_df = (df.isna().sum().sort_values(ascending  = False)/len(df))*100
```

```
percentage_missing_df
```

```
percentage_missing_df = percentage_missing_df[percentage_missing_df != 0]
```

```
percentage_missing_df.plot.barh()
```

```
#unique city counts
len(df.City.unique())
```

```
sorted_cities = df.City.sort_values(ascending=False).value_counts()[:20]
```
